```python
import sqlite3
import pandas as pd
import os

BASE_DIR = os.getcwd()
DB_PATH = os.path.join(BASE_DIR, "nstar.db")

conn = sqlite3.connect(DB_PATH)

print("Connected to NSTAR database.")
```

```python
inventory = pd.read_sql_query("""
    SELECT
        p.sku AS "SKU",
        p.product_name AS "Product Name",
        p.category AS "Category",
        i.quantity AS "Quantity",
        i.minimum_stock AS "Minimum Stock",
        i.last_updated AS "Last Updated"
    FROM products p
    JOIN inventory i
        ON p.product_id = i.product_id
    ORDER BY p.product_id
""", conn)

inventory
```

```python
transactions = pd.read_sql_query("""
    SELECT
        t.transaction_id AS "ID",
        p.sku AS "SKU",
        p.product_name AS "Product Name",
        t.action AS "Action",
        t.quantity_change AS "Change",
        t.previous_quantity AS "Previous Qty",
        t.new_quantity AS "New Qty",
        t.source AS "Source",
        t.device_id AS "Device",
        t.timestamp AS "Timestamp"
    FROM transactions t
    JOIN products p
        ON t.product_id = p.product_id
    ORDER BY t.transaction_id DESC
""", conn)

transactions
```

```python
cursor = conn.cursor()

cursor.execute("""
    UPDATE products
    SET rfid_uid = ?
    WHERE sku = ?
""", ("83F74F07", "NST-001"))

conn.commit()

print("RFID card assigned to NST-001.")
```

```python
rfid_assignments = pd.read_sql_query("""
    SELECT
        sku AS "SKU",
        product_name AS "Product Name",
        rfid_uid AS "RFID UID"
    FROM products
    ORDER BY product_id
""", conn)

rfid_assignments
```
